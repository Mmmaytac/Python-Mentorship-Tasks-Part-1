Task 5: Basic Dictionary Operations
-----------------------------------------------------------------------------

Create a dictionary:

Create a dictionary called student with the following key-value pairs:
"name": "Alice"
"age": 25
"gender": "Female"
Print the dictionary.
Access elements:

Print the value associated with the key "name" in the student dictionary.
Print the value associated with the key "age".
Modify elements:

Change the value associated with the key "age" to 26.
Print the modified dictionary.
Add and remove elements:

Add a new key-value pair "grade": A to the dictionary.
Remove the key "gender" from the dictionary.
Print the dictionary after each operation.

```python
# Create a dictionary
student = {
    "name": "Alice",
    "age": 25,
    "gender": "Female"
}
print("Original dictionary:", student)

# Access elements
print("Name:", student["name"])  # Output: Alice
print("Age:", student["age"])  # Output: 25

# Modify elements
student["age"] = 26
print("Modified dictionary:", student)  # Output: {'name': 'Alice', 'age': 26, 'gender': 'Female'}

# Add and remove elements
student["grade"] = "A"
print("After adding grade:", student)  # Output: {'name': 'Alice', 'age': 26, 'gender': 'Female', 'grade': 'A'}

del student["gender"]
print("After removing gender:", student)  # Output: {'name': 'Alice', 'age': 26, 'grade': 'A'}
```

Task 3: if_elsif
-----------------------------------------------------------------------------

Write a Python program that:

Asks the user to enter their weight (in kilograms) and height (in meters).
Calculates the Body Mass Index (BMI) using the formula: 
BMI = weight/height**2

Prints out the BMI value and classifies it based on the following criteria:
Underweight: BMI less than 18.5
Normal weight: BMI 18.5 to 24.9
Overweight: BMI 25 to 29.9
Obesity: BMI 30 or greater

Example Output:

Enter your weight (kg): 70
Enter your height (m): 1.75
Your BMI is 22.86. (Normal weight)


```python
weight = float(input("Enter your weight (kg): "))
height = float(input("Enter your height (m): "))

BMI = weight / height**2

if BMI < 18.5:
    bmi_class = 'Underweight'
elif 18.5 <= BMI < 25:
    bmi_class = 'Normal weight'
elif 25 <= BMI < 30:
    bmi_class = 'Overweight'
else:
    bmi_class = 'Obesity'

print(f"Your BMI is {BMI:.2f}. ({bmi_class})")
```

#This function calculates the remainder of the division of
#two numbers using keyword arguments and returns a detailed message about the result.


```python
def calculate_remainder(num1=12, num2=6):
    remainder = num1 % num2
    if remainder % 2 == 0:
        message = f"Remainder with keyword arguments is even: {remainder}"
    else:
        message = f"Remainder with keyword arguments is odd: {remainder}"
    return message

result = calculate_remainder(15, 7)  

# Correct way to print the result
print(result)
```

Calculate the Remainder with Keyword Arguments and Return a Detailed Message
This function calculates the remainder of the division of two 
numbers using keyword arguments and returns a detailed message about the result.


```python

def calculate_remainder_kw(num1=12, num2=7):
    remainder = num1 % num2
    if remainder % 2 == 0:
        message = f"Remainder with keyword arguments: {remainder} (even)"
    else:
        message = f"Remainder with keyword arguments: {remainder} (odd)"
    return message

# Example usage
result_kw = calculate_remainder_kw(num1=12, num2=7)
print(result_kw)
```
