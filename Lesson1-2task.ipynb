## Task1: Student Profile
- Get the input using input()
- Split the input using split()
- Write the name in uppercase
- Convert the age to int
- Store the languages as a list
- Find the number of languages
- Print the result using an f-string

```python
# input: Aysel, 18, Python Java SQL
# output:
# STUDENT: AYSEL
# AGE: 18
# SKILLS: ['Python', 'Java', 'SQL']
# TOTAL SKILLS: 3

info=input('melumati daxil edin: ')

lst=info.split(', ')
name=lst[0].upper()
age=int(lst[1])
language=lst[2].split(' ')
count=len(language)

print(f"STUDENT: {name}")
print(f"AGE: {age}")
print(f"skills: {language}")
print(f"TOTAL SKILLS: {count}")
```

## Task 2: Sentence Analyzer

- Convert the sentence to lowercase using lower()
- Split the sentence into words using split()
- Add the words to a list
- Find the longest word
- Store the length of each word in a dictionary


```python
# input :Python is very useful language
# Output:
# Word count: 5
# Longest word: language
# Word lengths: {'python': 6, 'is': 2, 'very': 4, 'useful': 6, 'language': 8}

info=input('infonu daxil edin: ').lower()
lst=info.split(' ')
word_len={}
longest_word=""

for word in lst:
    word_len[word]=len(word)
    if len(word)>len(longest_word):
        longest_word=word

print(f"Word count: {len(lst)}")
print(f"Longest word: {longest_word}")
print(f"Word lengths: {word_len}") 
    


```

## Task 3: Product Price Calculator

- prices = {"apple": 1.5, "banana": 1, "orange": 2}
- Split the products using split()
- Create a dictionary
- Calculate the total price
- Print the result for each product using formatted output


```python
# Input example:
# apple:2, banana:3, orange:1
# Output:
# apple -> 2 x 1.5 = 3.0
# banana -> 3 x 1 = 3
# orange -> 1 x 2 = 2
# Total price: 8.0 AZN

info=input()
prices = {"apple": 3, "banana": 5, "orange": 2.7}
lst=info.split(', ')

total=0
for item in lst:
    name, qty=item.split(':')
    qty=int(qty)
    cost=qty*prices[name]
    total+=cost
    print(f" {name} -> {qty} X {prices[name]} = {cost}")

print(f"Total price: {total} AZN")
```

## Task 4: Exam Result Analysis

- Split the data using split()
- Create a dictionary
- Calculate the average score
- Add students who scored above 85 to a list
- Find the student with the highest score


```python
# Input example:
# Ali=78, Nigar=92, Murad=65, Aysel=88
# Output:
# Average score: 80.75
# Top students: ['Nigar', 'Aysel']
# Highest score: Nigar - 92

info=input()
lst=info.split(", ")
scores={}

for i in lst:
    name, score=i.split('=')
    scores[name]=int(score)

average=sum(scores.values())/len(scores)

top_students=[]
highest_score=0
highest_name=""

for name, score in scores.items():
    if score > 85:
        top_students.append(name)
    if score > highest_score:
        highest_score = score
        highest_name = name


print(f" Average score: {average}")
print(f"Top students: {top_students}")
print(F"Highest score: {highest_name} - {highest_score}")      


```

## Task6: List Processing

- Create a list:
[5, 12, 8, 5, 23, 12, 8, 15, 23, 5]
- Remove duplicate elements using set
- Keep only even numbers in the new list
- Sort the result from largest to smallest
- Expected output:
[12, 8]

```python
lst= [5, 12, 8, 5, 23, 12, 8, 15, 23, 5]
unikal=list(set(lst))

even_num=[]
for i in unikal:
    if i%2==0:
        even_num.append(i)

even_num.sort(reverse=True)
print(even_num)
```

## Task 7: 2-Digits Counter

- Use a for loop to go through the list
- Use if to count numbers between 10 and 99
- Use a while loop to calculate the sum of those numbers
- Print the results

```python
numbers=[11,234,4,33,456,4,34,66]
lst=[]
for i in numbers:
    if 10<=i<=99:
        lst.append(i)
i=0
total=0
while i<len(lst):
    total+=lst[i]
    i+=1

print(lst)
print(total)
```
