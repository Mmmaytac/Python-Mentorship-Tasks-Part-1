```python
numbers=[2,4,32,5,67]
netice=list(map(lambda x: x*3,numbers))
print(netice)
```

```python
numbers=[2,4,32,5,67]
netice=list(filter(lambda x: x%2==0,numbers))
print(netice)
```

```python
from functools import reduce
numbers=[2,4,3,5,6]
netice=reduce(lambda x,y: x*y,numbers)
print(netice)
```

```python
numbers=list(range(1,10))
odd_number=list(filter(lambda x: x%2!=0, numbers))
squared_odd_num=list(map(lambda x: x**2, odd_number))
print(odd_number)
print(squared_odd_num)
```
